```python
def count_vowel(str):
    vowel = 0
    for c in str:
        if c in ('A', 'E', 'I', 'O', 'U', 'a', 'e', 'i', 'o', 'u'):
            vowel = vowel+1
    return vowel
```

```python
def area(width, height):
    c = width * height
    return c
```

```python
def factorial(num) :
    if num == 0 :
        return 1
    else :
        return num * factorial(num-1)
print(factorial(5))
```
