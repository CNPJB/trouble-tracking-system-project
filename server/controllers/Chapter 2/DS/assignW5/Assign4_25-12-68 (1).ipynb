```python
string = input() 
long = 0 
ans = ""
for i in range(0 , len(string)) : 
    temp_ans = [] 
    temp_set = set()
    for j in range(i,len(string)) : 
        if string[j] not in temp_set : 
            temp_ans.append(string[j])
            temp_set.add(string[j])
        else : 
            if len(temp_ans) > long : 
                long = len(temp_ans) 
                ans = str(temp_ans)
        
```
