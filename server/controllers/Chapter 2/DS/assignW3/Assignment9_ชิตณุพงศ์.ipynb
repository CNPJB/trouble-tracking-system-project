```python
#Assignment9
exs = int(input("Enter Exercise Score: "))
mid = int(input("Enter Midterm Exam Score: "))
fnl = int(input("Enter Final Exam Score: "))

score = exs + mid + fnl

if ( score >= 80 ):
    grade = "A"
elif ( score >= 75 ):
    grade = "B+"
elif ( score >= 70 ):
    grade = "B"
elif ( score >= 65 ):
    grade = "C+"
elif ( score >= 60 ):
    grade = "C"
elif ( score >= 55):
    grade = "D+"
elif ( score >= 50):
    grade = "D"
else:
    grade = "F"
    
if ( grade == "A" ):
    status = "ผ่านเกณฑ์ระดับดีมาก"
elif ( grade == "B+" or grade == "B"):
    status = "ผ่านเกณฑ์ระดับดี"
elif ( grade == "C+" or grade == "C"):
    status = "ผ่านเกณฑ์ระดับพอใช้"
elif ( grade == "D+" or grade == "D"):
    status = "ผ่านเกณฑ์ระดับควรปรับปรุง"
elif ( grade == "F" ):
    status = "ไม่ผ่าน"

print("====Result====")
print("Your Score Is: %d\nGrade: %s\nStatus: %s" %(score, grade, status))
```
